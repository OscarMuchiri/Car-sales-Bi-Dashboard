```python
# Loading thedataset
car_data = pd.read_csv('Car_sales.csv')
```

```python
# Initial Data Exploration
print("Dataset Shape:", car_data.shape)
print("Columns:", car_data.columns)
car_data.head()
```

```python
# Handle Missing Values and Invalid Entries
# Replace placeholders like '.' or 'NA' with NaN
car_data.replace(['.', 'NA', 'N/A', 'na'], np.nan, inplace=True)

# Convert applicable columns to numeric data types
numeric_cols = [
    '4-year resale value', 'Price in thousands', 'Engine size', 'Horsepower',
    'Wheelbase', 'Width', 'Length', 'Curb weight', 'Fuel capacity', 'Fuel efficiency'
]

for col in numeric_cols:
    car_data[col] = pd.to_numeric(car_data[col], errors='coerce')
```

```python
# Handle Missing Numeric Values by Filling with Median
for col in numeric_cols:
    car_data[col] = car_data[col].fillna(car_data[col].median())
```

```python
# Creating Price Category Column
car_data['Price Category'] = pd.cut(
    car_data['Price in thousands'],
    bins=[0, 20, 35, np.inf],
    labels=['Low', 'Medium', 'High'])
```
